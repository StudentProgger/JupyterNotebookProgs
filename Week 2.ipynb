## Why we are using series instead of list or dictionary

In Python, the choice between a list, dictionary, or a Pandas Series depends on the specific use case and the nature of the data that you are working with.
Structured Data: Series and DataFrames are designed to work together  with structured data. They provide more structure to your data compared to list and dictionaries.
Vectorized Operations: Series will support vectorized operations, what can be more efficient than operations in lists or dictionaries.
Data Indexing: Series will support both integer-based and label-based indexing. This can make data manipulation easier.
Handling Missing Data: Series have built-in functionality for handling missing data.
