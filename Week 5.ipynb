## Import all the necessary libraries

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.svm import SVC
from sklearn.metrics import precision_score, recall_score, accuracy_score, f1_score
```

## Identify features and class variables

```python
app_df = pd.read_pickle('/Users/Vladgappasov/Downloads/Android_malware1.pkl')
app_df.head()
```

```python
X = app_df.iloc[:, 8:]
Y = app_df['malware']
```

## Split the data to training and test set

```python
# Split the data into training and testing sets
X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size=0.2, random_state=42)
```

## Model


```python
# Create an SVM classifier object
svm_classifier = SVC()
#Train the SVM classifier
svm_classifier.fit(X_train, Y_train)
#Make predictions on the test set.
Y_pred= svm_classifier.predict(X_test)
```

## Model SVM with various kernels

```python
svm_classifier_poly= SVC(kernel='poly') #polynomial
```

## Evaluate

```python
#Evaluate the accuracy of the classifier
accuracy= accuracy_score(Y_test, Y_pred)
print("Accuracy", accuracy)
```

```python
#Evaluate the precision of the classifier
precision= precision_score(Y_test, Y_pred)
print("Precision", precision)
```

```python
#Evaluate the recall of the classifier
recall= recall_score(Y_test, Y_pred)
print("Recall", recall)
```

```python
#Evaluate the f-measure of the classifier.
f_measure = f1_score(Y_test, Y_pred)
print("f_measure:", f_measure)
```

## Evaluate the other classifiers with kernels

```python
#Train the SVM classifier
svm_classifier_poly.fit(X_train, Y_train)
#Make predicitions based on the test set
Y_pred_poly= svm_classifier_poly.predict(X_test)
#Evaluate the accuracy of the classifier
accuracy_poly= accuracy_score(Y_test, Y_pred_poly)
print("Accuracy", accuracy_poly)
```
