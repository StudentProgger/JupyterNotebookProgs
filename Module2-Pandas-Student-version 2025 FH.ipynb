# Python Dictionaries
Dictionaries are used to store data values in key:value pairs.

A dictionary is a collection which is ordered, changeable and does not allow duplicates.


<br>

```python
thisdict = {
  "Brand": "Ford",
  "Model": "Mustang",
  "Year": 1964
}
x = thisdict["Model"] # get item
thisdict["Year"]="2000" #update item
thisdict["Color"] = "red" # Add item
del thisdict["Model"] #Remove item
#access item
for x in thisdict.values():
    print(x)  #notice that we don't see the model anymore
```

# Series Data Structure

A Pandas Series is like a column in a table.

It is a one-dimensional array holding data of any type.

```python
import pandas as pd
```

You can also use a key/value object, like a dictionary, when creating a Series.

```python
d = {'a': 1, 'b': 2, 'c': 3}
ser = pd.Series(data=d, index=['a', 'b', 'c'])
ser
```

# Why we are using series instead of list or dictionary

In Python, the choice between using a list, dictionary, or a Pandas Series depends on the specific use case and the nature of the data you are working with. 

Structured Data: Series and DataFrames are designed to work with structured data. They provide more structure to your data compared to lists and dictionaries. 

Vectorized Operations:Series support vectorized operations, which can be more efficient than operations on lists or dictionaries.

Data Indexing: Series supports both integer-based and label-based indexing. This can make data manipulation easier.

Handling Missing Data: Series have built-in functionality for handling missing data.

## <p style="color:red;">Exercise 6:

<p style="color:red;"> Extract number of rows modified between May 5,2022 to June 5,2022?
    Extract number of distinct cves reported between March 5,2022 to April 5,2022?


```python
import pandas as pd
df = pd.read_csv("/Users/vladgappasov/Downloads/nvd_cve_2022.csv")
df['last_modified_date'] = pd.to_datetime(df['last_modified_date'], errors='coerce', utc=True)

# Define UTC-aware datetime range
start_mod = pd.to_datetime('2022-05-05', utc=True)
end_mod = pd.to_datetime('2022-06-05', utc=True)

# Filter rows by last_modified_date
modif_rows = df[(df['last_modified_date'] >= start_mod) & (df['last_modified_date'] <= end_mod)]

# Count number of rows
num_mod_rows = len(modif_rows)
print(f"Number of rows modified between May 5 and June 5, 2022: {num_mod_rows}")
```

```python
import pandas as pd

df = pd.read_csv("/Users/vladgappasov/Downloads/nvd_cve_2022.csv")
df['publish_date'] = pd.to_datetime(df['publish_date'], errors='coerce', utc=True)
start_date = pd.to_datetime('2022-03-05', utc=True)
end_date = pd.to_datetime('2022-04-05', utc=True)
published_rows = df[(df['publish_date'] >= start_date) & (df['publish_date'] <= end_date)]
num_distinct_cves = published_rows['id'].nunique()
print(f"Number of distinct CVEs published between March 5 and April 5, 2022: {num_distinct_cves}")
```
